## Часть 1 - API
С помощью API openweathermap получите информацию о погоде (достаточный список параметров определите сами) в 5 самых старых их ныне существующих городов (используйте нейросеть для поиска информации про эти города). В качестве ответа предоставьте код и вывод на экран, отвечающий на вопрос задачи.

```python
import requests


API_KEY = "c86250163953d321b9b61edeb46ad4c7"
cities = ['Jericho', 'Byblos', 'Damascus', 'Aleppo', 'Athens']


def get_weather(city):
    url = "http://api.openweathermap.org/data/2.5/weather"

    params = {
        "q": city,
        "appid": API_KEY,
        "units": "metric",
        "lang": "ru"
    }

    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    return response.json()


def print_weather(data):
    city_name = data.get("name", "Неизвестный город")

    main = data.get("main", {})
    temperature = main.get("temp")
    feels_like = main.get("feels_like")
    humidity = main.get("humidity")

    weather_list = data.get("weather", [])

    if weather_list:
        description = weather_list[0].get("description", "Описание отсутствует")
    else:
        description = "Описание отсутствует"

    wind = data.get("wind", {})
    wind_speed = wind.get("speed")

    print()
    print("Город:", city_name)
    print("Температура:", temperature, "°C")
    print("Ощущается как:", feels_like, "°C")
    print("Влажность:", humidity, "%")
    print("Погода:", description)
    print("Скорость ветра:", wind_speed, "м/с")
    print('-' * 60)


def main():
    for city in cities:

      try:
          weather = get_weather(city)
          print_weather(weather)

      except requests.exceptions.Timeout:
          print("Ошибка: сервер слишком долго не отвечает.")

      except requests.exceptions.ConnectionError:
          print("Ошибка: нет соединения с сервером.")

      except requests.exceptions.HTTPError as error:
          print("HTTP-ошибка:", error)

      except requests.exceptions.JSONDecodeError:
          print("Ошибка: сервер вернул некорректный JSON.")

      except requests.exceptions.RequestException as error:
          print("Ошибка запроса:", error)


main()
```

## Часть 2 - API
1. Выберите любой понравившийся сайт
2. С помощью BeautifulSoup выведите на экран все основные элементы сайта (заголовки, ссылки, списки (если есть), таблицы (если есть) и тд)
3. В качестве ответа предоставьте код и вывод на экран, отвечающий на вопрос задачи

```python
import requests
from bs4 import BeautifulSoup

url = 'https://boxx.ru/'
headers = {
    'User-Agent': 'Mozilla/5.0'
}
try:
  response = requests.get(url, headers=headers, timeout=10)
  response.raise_for_status()
  print('Код ответа', response.status_code)
  soup = BeautifulSoup(response.content, 'html.parser')

  # Найдем заголовок страницы
  title = soup.find('title')
  if title:
      print()
      print('Заголовок вкладки:')
      print(title.get_text(strip=True))
      print("*" * 130)
      print("*" * 130)
      print("*" * 130)
      print("*" * 130)
  else:
      print()
      print("Тег title не найден")

  # Найдем все заголовки на странице
  titles_h1 = soup.find_all("h1")
  if titles_h1:
      print()
      print('Все заголовки на странице:')
      for title in titles_h1:
        print(title.get_text(strip=False))
      print("*" * 130)
      print("*" * 130)
      print("*" * 130)
      print("*" * 130)
  else:
      print()
      print("Тег h1 не найден")

  # Найдем все ссылки на странице
  links = soup.find_all("a")
  if links:
    print()
    print('Ссылки:')
    print("Количество ссылок:", len(links))

    for link in links[:10]:
      text = link.get_text(strip=True)
      href = link.get("href")

      print("Текст:", text)
      print("Адрес:", href)
  else:
    print()
    print('Ссылки на странице отсутствуют')
  print('*' * 130)
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)

  # Найдем все ненумированные списки на странице
  u_lists = soup.find_all('ul')
  if u_lists:
    print()
    print('Ненумированные списки:')
    for ul in u_lists:
      lists = ul.find_all('li', recursive=False)
      for li in lists:
        print(li.get_text(strip=True))
      print('-' * 30)
  else:
    print()
    print('Ненумированные списки на странице отсутствуют')
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)

  # Найдем все параграфы на странице
  paragraphs = soup.find_all('p')
  if paragraphs:
    print()
    print('Параграфы:')
    for p in paragraphs:
      print(p.get_text(strip=False))
      print('-' * 30)
  else:
    print()
    print('Параграфы на странице отсутствуют')
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)

  # Найдем все нумированные списки на странице
  o_lists = soup.find_all('ol')
  if o_lists:
    print()
    print('Нумированные списки:')
    for ol in o_lists:
      lists = ol.find_all('li', recursive=False)
      for li in lists:
        print(li.get_text(strip=True))
      print('-' * 30)
  else:
    print()
    print('Нумированные списки на странице отсутствуют')
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)

  # Найдем все кнопки на странице
  buttons = soup.find_all('button')
  if buttons:
    print()
    print('Кнопки:')
    for button in buttons:
      print(button.get_text(strip=False))
  else:
    print()
    print('Кнопки на странице отсутствуют')
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)

  # Найдем таблицу на странице
  table = soup.find('table')
  if table:
    print()
    print('Таблица:')
    rows = table.find_all('tr')
    for row in rows:
      cells = row.find_all(['td', 'th'])
      row_data = [cell.get_text().strip() for cell in cells]
      print(', '.join(row_data))
  else:
    print()
    print('Таблица на странице отсутствует')
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)
  print("*" * 130)

except requests.exceptions.HTTPError as err:
    print(f"Ошибка HTTP: {err}")
except requests.exceptions.RequestException as err:
    print(f"Произошла ошибка: {err}")
```
