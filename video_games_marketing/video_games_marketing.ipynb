## Задача 2. Очистка данных

- Приведите названия столбцов к нижнему регистру.
- В `user_score` замените  `'tbd'` на `NaN` и объясните, что оно означает.
- Приведите `user_score` к числовому типу, `year_of_release` — к целочисленному. (подсказка: для замены поможет метод `.replace`, для создания NaN - библиотека `import numpy as np` и `np.nan`, для преобразования к числовому типу используйте типы данных с поддержкой NaN - Int64, Float32 и тд)

###Решение

```python
# Приведем названия столбцов к нижнему регистру

df.columns = (col.lower() for col in df.columns)
```

```python
# Проверим, что названия столбцов стали в нижнем регистре. Для этого выведем первые три строки

df.head(3)
```

```python
# В user_score заменим 'tbd' на NaN

df['user_score'] = df['user_score'].replace('tbd', np.nan)
```

```python
# Проверим, что 'tbd' заменилось на NaN. Для этого выведем все уникальные значения столбца user_score

df['user_score'].unique()
```

'tbd' расшифровывается как: To Be Determined или To Be Decided - в переводе 'будет определено позже' или 'будет решено позже'. В датасетах это просто текстовая заглушка и ее лучше заменить на NaN для дальнейшей работы с пропусками

```python
# Приведем user_score к числовому типу

df['user_score'] = df['user_score'].astype('Float64')

# Приведем year_of_release к целочисленному типу

df['year_of_release'] = df['year_of_release'].astype('Int64')
```

```python
# Проверим, что типы данных в столбцах поменялись. Для этого выведем информацию о типах данных во всей таблице

df.info()
```

## Задача 5. Анализ платформ

###Решение

```python
# Выведем таблицу: платформа — суммарные продажи — количество игр — медианные продажи на игру
df_platform = df.groupby('platform', as_index=False).agg(
    sum_sales=('total_sales', 'sum'),
    number_of_games=('name', 'count'),
    median_sales=('total_sales', 'median'))
df_platform
```

## Задача 6. Анализ жанров

Для каждого жанра рассчитайте суммарные и медианные продажи, количество игр. Выведите таблицу по убыванию суммарных продаж. Определите самый прибыльный жанр и жанр с наибольшими медианными продажами.

###Решение

```python
# Для каждого жарна рассчитаем суммарные и медианные продажи, количество игр. Выведем таблицу по убыванию суммарных продаж
df_genre = df.groupby('genre', as_index=False).agg(
    sum_sales=('total_sales', 'sum'),
    median_sales=('total_sales', 'median'),
    number_of_games=('name', 'count')
)
df_genre.sort_values('sum_sales', ascending=False)
```

```python
# Определим самый прибыльный жанр
top_genre = df_genre.loc[df_genre['sum_sales'].idxmax(), 'genre']
print('Самый прибыльный жанр:', top_genre)
```

```python
# Определим жанр с наибольшими медианными продажами
top_genre = df_genre.loc[df_genre['median_sales'].idxmax(), 'genre']
print('Жанр с наибольшими медианными продажами:', top_genre)
```

## Задача 7. Региональные предпочтения: жанры и рейтинг

- Для каждого региона выведите топ-5 жанров по продажам.
- Рассчитайте суммарные продажи по рейтингу ESRB в каждом регионе. Какой рейтинг лидирует?

###Решение

```python
# Для каждого региона выведем топ-5 жанров по продажам

# Создадим таблицу в которой будет не 4 столбца с разными регионами, а один столбец с четырьмя регионами
cols = ['na_sales', 'eu_sales', 'jp_sales', 'other_sales']
df_melted = df.melt(
    id_vars=['genre'],
    value_vars=cols,
    var_name='region',
    value_name='sales'
)

# Суммируем продажи по регионам и рейтингу
df_melted_sum = df_melted.groupby(['region', 'genre'], as_index=False)['sales'].sum()

# Заменим названия регионов на более понятные
regions = {'eu_sales': 'EU', 'jp_sales': 'Japan', 'na_sales': 'North America', 'other_sales': 'Other countries'}
df_melted_sum['region'] = df_melted_sum['region'].replace(regions)

# Выведем топ-5 жанров по продажам в регионах
df_melted_sum.sort_values(['region', 'sales'], ascending=[True, False]).groupby('region', group_keys=False).head()
```

Рассчитаем суммарные продажи по рейтингу ESRB в каждом регионе. Выведем таблицу с лидерами по рейтингу в каждом регионе

```python
# Проверим есть ли пропуски в столбце rating
df['rating'].isna().sum()
```

```python
# Заменим NaN на Unknown
df['rating'] = df['rating'].fillna('Unknown')

# Создадим таблицу в которой будет не 4 столбца с разными регионами, а один столбец с четырьмя регионами
cols = ['na_sales', 'eu_sales', 'jp_sales', 'other_sales']
df_melted = df.melt(
    id_vars=['rating'],
    value_vars=cols,
    var_name='region',
    value_name='sales'
)

# Суммируем продажи по регионам и рейтингу
df_melted_sum = df_melted.groupby(['region', 'rating'], as_index=False)['sales'].sum()

# Найдем индексы строк с максимальными продажами
leaders = df_melted_sum.groupby('region')['sales'].idxmax()

# Получим таблицу с лидерами по рейтингу
df_leaders = df_melted_sum.loc[leaders].sort_values('region')

# Заменим названия регионов на более понятные и выведем конечную таблицу
regions = {'eu_sales': 'EU', 'jp_sales': 'Japan', 'na_sales': 'North America', 'other_sales': 'Other countries'}
df_leaders['region'] = df_leaders['region'].replace(regions)
df_leaders
```
