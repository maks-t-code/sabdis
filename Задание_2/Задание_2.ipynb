## Задача 7. Средняя цена по городам

Посчитайте среднюю цену квартиры для каждого города.

**Сделайте вывод:**

- В каком городе самая высокая средняя цена?
- В каком городе самая низкая средняя цена?
- Насколько сильно отличаются цены между городами?

````python
df.groupby('city')['price_rub'].mean().reset_index().sort_values('price_rub', ascending=False)
````

````python
# Создадим датасет средней цены по городам
df_mean = df.groupby('city')['price_rub'].mean()

# Найдем в каком городе самая высокая средняя цена
print('Самая высокая средняя цена в городе:', df_mean.idxmax())

# Найдем в каком городе самая низкая средняя цена
print('Самая низкая средняя цена в городе:', df_mean.idxmin())

# Найдем разницу в средней цене между этими городами
price_range = df_mean.max() - df_mean.min()
print('Разница в средней цене на квартиры между этими городами составляет:', price_range)
````

###Вывод
Посчитали среднюю цену квартиры для каждого города. Оказалось, что самая высокая средняя цена на квартиру не в Москве, а в Санкт-Петербурге и составляет 18 260 000 рублей. Самая низкая средняя цена в Самаре, и она равна 4 660 000 рублей. Разница между ними составляет 13 600 000 рублей. Это может указывать на то, что в нашем датасете представлены только очень дорогие квартиры из Санкт-Петербурга. Для более объективного анализа нам необходимо иметь более сбалансированную выборку по количеству и ценам на квартиры в каждом городе
