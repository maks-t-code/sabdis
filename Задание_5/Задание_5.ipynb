## Задача 2. Работа с датой регистрации

Преобразуйте столбец `registration_date` к типу даты.

Добавьте новые столбцы:

- `registration_weekday` — день недели регистрации;
- `registration_month` — месяц регистрации.

Постройте гистограмму количества регистраций по месяцам.

````python
# Преобразуем столбец registration_date к типу даты

data['registration_date'] = pd.to_datetime(data['registration_date'])
````

````python
# Проверим, что тип данных столбца registration_date изменился на datetime
data.dtypes
````

````python
# Добавим новые столбцы:
# registration_weekday — день недели регистрации;
# registration_month — месяц регистрации.

data['registration_weekday'] = data['registration_date'].dt.weekday
data['registration_month'] = data['registration_date'].dt.month
````

````python
# Проверим, что новые столбцы добавились в датасет
data.head(3)
````

````python
# Построим гистограмму количества регистраций пользователей по месяцам

sns.histplot(data=data, x='registration_month', discrete=True, color='skyblue')
plt.title('Количество регистраций пользователей по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Количество регистраций пользователей')
plt.show()
````

Вывод: как видно из графика, в представленном датасете есть данные только за полгода. Наибольшее количество регистраций пользователей было в марте и в мае

## Задача 4. Анализ прохождения обучения

Создайте новый столбец:

```text
completion_rate
```

Он должен показывать долю пройденных уроков:

```text
lessons_completed / lessons_total
```

После этого:

1. Постройте гистограмму `completion_rate`.
2. Найдите среднее и медиану `completion_rate`.
3. Сравните средний `completion_rate` для тех, кто получил сертификат, и тех, кто не получил.

````python
# Создадим новый признак completion_rate

data['completion_rate'] = data['lessons_completed'] / data['lessons_total']
````

````python
# Проверим, что новый признак есть в датасете

data.head(3)
````

````python
# Построим гистограмму completion_rate

sns.histplot(data=data, x='completion_rate')
plt.title('Распределение доли пройденных уроков')
plt.xlabel('Доля пройденных уроков')
plt.ylabel('Количество пользователей')
plt.show()
````

Вывод: видим, что большинство пользователей завершило большую часть уроков, но также есть группа пользователей с низкой долей завершенности

````python
# Посчитаем среднее и медиану completion_rate и выведем значение в процентах

print('Среднее значение completion_rate:', round(data['completion_rate'].mean(), 3) * 100, '%')
print('Медианное значение completion_rate:', round(data['completion_rate'].median(), 3) * 100, '%')
````

Вывод: средний уровень завершенности уроков составляет 75.2%. Медианное значение составляет 81.9%. Это говорит о том, что половина пользователей завершила более 80% уроков

````python
# Сравним средний completion_rate для тех, кто получил сертификат, и тех, кто не получил

data.groupby('certificate_received')['completion_rate'].mean().reset_index()
````

Вывод: у тех, кто получил сертификат средний уровень завершенности уроков составляет 90.4%. А те, кто не получил сертификат, в среднем завершили 49.7% уроков

## Задача 5. Поиск закономерностей в продажах

С помощью сводных таблиц ответьте на вопросы:

1. Какая категория курсов приносит больше всего выручки по `final_price`?
2. У какого источника привлечения (`lead_source`) самая высокая средняя итоговая цена покупки?
3. У какого курса самая высокая средняя удовлетворённость `satisfaction_score`?
4. Постройте график зависимости медианной итоговой цены от месяца регистрации.
5. Сделайте 2–3 вывода по результатам анализа.

````python
# Составим сводную таблицу и выясним какая категория курсов приносит больше всего выручки

data_course_category = data.pivot_table(
    index='course_category',
    values='final_price',
    aggfunc='sum'
).sort_values('final_price', ascending=False).reset_index()
data_course_category.columns = ['Категория курсов', 'Выручка']
data_course_category
````

Вывод: больше всего выручки приносит категория Программирование: 1578900

````python
# Составим сводную таблицу и выясним у какого источника привлечения (lead_source) самая высокая средняя итоговая цена покупки

data_lead_source = data.pivot_table(
    index='lead_source',
    values='final_price',
    aggfunc='mean'
).sort_values('final_price', ascending=False).reset_index()
data_lead_source.columns = ['Источник привлечения', 'Средняя итоговая цена']
data_lead_source
````

Вывод: Самая высокая средняя итоговая цена покупки у источника привлечения Рекомендация: 53171.428571

````python
# Составим сводную таблицу и выясним у какого курса самая высокая средняя удовлетворённость satisfaction_score

data_course = data.pivot_table(
    index='course_name',
    values='satisfaction_score',
    aggfunc='mean'
).sort_values('satisfaction_score', ascending=False).reset_index()
data_course.columns = ['Название курса', 'Средняя оценка удовлетворенности']
data_course
````

Вывод: самая высокая средняя оценка удовлетворенности у курса Project Management: 7.94

````python
# Составим сводную таблицу и выясним зависимость медианной итоговой цены от месяца регистрации

data_registration_month = data.pivot_table(
    index='registration_month',
    values='final_price',
    aggfunc='median'
).reset_index()
data_registration_month
````

````python
# Построим график

plt.plot(data_registration_month['registration_month'], data_registration_month['final_price'], marker='o')
plt.title('Зависимость медианной итоговой цены от месяца регистрации')
plt.xlabel('Месяц регистрации')
plt.ylabel('Медианная итоговая цена')
plt.show()
````

Вывод: анализ медианной итоговой цены в зависимости от месяца регистрации пользователя выявил стабильность ценовой политики. Значения варьируются в узком диапазоне от 45600 до 48800. Максимальное значение зафиксировано в марте (48800), минимальное - в феврале(45600). Поскольку данные представлены всего за полгода, то нельзя со стопроцентной уверенностью утверждать об отсутствии сезонности. Однако из представленных данных можно сделать вывод, что сезонность не оказывает существенного влияния на стоимость покупки курса
